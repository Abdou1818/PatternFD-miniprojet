# Étapes 2+3 — Génération des candidats

Pour chaque pattern sur col_X, on regarde la valeur la plus fréquente de col_Y.
Si c'est assez fréquent -> candidat PFD.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
from src.pattern_extractor import extract_all_patterns
from src.candidate_generator import generate_all_candidates
```

```python
t2 = pd.read_csv('../data/pfd_validation/t2.csv')

# on va extraire les patterns pour toutes les colonnes
pattern_maps = {}
for col in t2.columns:
    pmap = extract_all_patterns(t2[col], k_max=3, min_support=30)
    if pmap:
        pattern_maps[col] = pmap

print(f'Colonnes avec patterns: {list(pattern_maps.keys())}')
```

```python
# générer les candidats
candidates = generate_all_candidates(t2, pattern_maps, min_support=30)
print(f'{len(candidates)} candidats générés')

# afficher les 20 premiers
for c in candidates[:20]:
    print(f"  {c['col_X']} [{c['match_type_X']}:'{c['pattern_X']}'] -> {c['col_Y']} = {c['pattern_Y']!r}  (conf brute: {c['raw_confidence']:.1%})")
```
