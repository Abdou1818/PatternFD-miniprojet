## t1.csv — Employés (9K lignes)

```python
t1 = pd.read_csv('../data/pfd_validation/t1.csv')
print('Shape:', t1.shape)
print('Colonnes:', list(t1.columns))
t1.head()
```

```python
# quels patterns on voit ?
# Full Name a un format 'Nom, Prénom' -> le premier mot donne le nom de famille
print(t1['Full Name'].head(10).tolist())
print('\nGenre:', t1['Gender'].unique())
print('Nb départements:', t1['Department'].nunique())
print('\nValeurs manquantes:')
print(t1.isnull().sum())
```

## t2.csv — Entreprises (3.5K lignes)

```python
t2 = pd.read_csv('../data/pfd_validation/t2.csv')
print('Shape:', t2.shape)
t2.head()
```

```python
t2['ZIP'].dropna().count()
```
